## 字符串的乐趣

```python
# bonjour
def is_palindrome(text):
    normalized = "".join(char.lower() for char in text if char.isalnum())
    return normalized == normalized[::-1]


def is_abecedarian(text):
    letters = text.replace(" ", "").lower()
    return all(left <= right for left, right in zip(letters, letters[1:]))


def remove_vowels(text):
    vowels = "aeiou"
    return "".join(char for char in text if char.lower() not in vowels)


assert is_palindrome("T A C O cat")
assert is_palindrome("Was it a car or a cat I saw?")
assert not is_palindrome("python")
assert is_abecedarian("abcd efgh xyz")
assert is_abecedarian("beet")
assert not is_abecedarian("dog")
assert remove_vowels("cat") == "ct"
assert remove_vowels("audio") == "d"
assert remove_vowels("AEIOU") == ""
print("回文判断（T A C O cat）：", is_palindrome("T A C O cat"))
print("回文判断（Was it a car or a cat I saw?）：", is_palindrome("Was it a car or a cat I saw?"))
print("字母顺序判断（abcd efgh xyz）：", is_abecedarian("abcd efgh xyz"))
print("去除元音（audio）：", remove_vowels("audio"))
print("字符串函数测试通过。")
```

`is_palindrome` 只保留字母和数字并统一大小写，再比较字符串与其逆序，因此会忽略空格和标点符号。`is_abecedarian` 去掉空格后逐对比较相邻字母，允许重复字母。`remove_vowels` 逐字符过滤元音，同时保留其他字符的原有顺序和大小写；全是元音时会返回空字符串。本代码单元开头的 `# bonjour` 是作业要求保留的注释。

## 列表的乐趣

```python
def list_reverse(items):
    if not isinstance(items, list):
        raise TypeError("输入必须是列表。")
    return items[::-1]


def is_sorted(seq):
    iterator = iter(seq)
    try:
        previous = next(iterator)
    except StopIteration:
        return True

    for current in iterator:
        if current < previous:
            return False
        previous = current
    return True


assert list_reverse([1, 2, 3]) == [3, 2, 1]
assert is_sorted([])
assert is_sorted([1, 2, 2, 5])
assert not is_sorted([1, 3, 2])
print("反转 [1, 2, 3]：", list_reverse([1, 2, 3]))
print("[1, 2, 2, 5] 是否升序：", is_sorted([1, 2, 2, 5]))
print("[1, 3, 2] 是否升序：", is_sorted([1, 3, 2]))
print("列表函数测试通过。")
```

`list_reverse` 先确认参数是列表，再用切片返回逆序的新列表，不会修改原列表；错误类型会触发 `TypeError`。`is_sorted` 只创建并遍历一次迭代器，通过相邻元素判断是否升序，空序列和只有一个元素的序列也视为有序。

## 作为向量的元组

```python
def _is_number(value):
    return isinstance(value, (int, float)) and not isinstance(value, bool)


def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("向量必须是元组。")
    if not _is_number(s):
        raise TypeError("标量必须是整数或浮点数。")
    if not all(_is_number(value) for value in t):
        raise TypeError("向量元素必须是整数或浮点数。")
    return tuple(value * s for value in t)


def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个向量都必须是元组。")
    if len(x) != len(y):
        raise ValueError("两个向量的长度必须相同。")
    if not all(_is_number(value) for value in x + y):
        raise TypeError("向量元素必须是整数或浮点数。")
    return float(sum(left * right for left, right in zip(x, y)))


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if not mx:
        return True

    column_count = len(mx[0])
    return all(
        len(row) == column_count and all(_is_number(value) for value in row)
        for row in mx
    )


def mx_vec_mult(mx, vector):
    if not isinstance(mx, tuple):
        raise TypeError("矩阵必须是元组的元组。")
    if not check_valid_mx(mx):
        raise ValueError("矩阵格式无效。")
    if not isinstance(vector, tuple):
        raise TypeError("向量必须是元组。")
    if not all(_is_number(value) for value in vector):
        raise TypeError("向量元素必须是整数或浮点数。")
    if mx and len(mx[0]) != len(vector):
        raise ValueError("矩阵的列数必须与向量长度相同。")
    return tuple(sum(a * b for a, b in zip(row, vector)) for row in mx)


assert vec_scalar_mult((1, 2.5), 2) == (2, 5.0)
assert vec_inner_product((1, 2, 3), (4, 5, 6)) == 32.0
assert check_valid_mx(((1, 2), (3.5, 4)))
assert not check_valid_mx(((1, 2), (3,)))
assert mx_vec_mult(((1, 2), (3, 4)), (5, 6)) == (17, 39)
print("(1, 2.5) 乘以 2：", vec_scalar_mult((1, 2.5), 2))
print("(1, 2, 3) 与 (4, 5, 6) 的内积：", vec_inner_product((1, 2, 3), (4, 5, 6)))
print("矩阵 ((1, 2), (3.5, 4)) 是否有效：", check_valid_mx(((1, 2), (3.5, 4))))
print("矩阵 ((1, 2), (3, 4)) 乘以向量 (5, 6)：", mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
print("元组向量和矩阵函数测试通过。")
```

`_is_number` 统一检查整数和浮点数，并排除布尔值。`vec_scalar_mult` 和 `vec_inner_product` 分别验证向量、标量类型以及向量长度，内积结果显式转换为浮点数。`check_valid_mx` 检查矩阵是否由等长的数值元组组成；空外层元组也满足题目列出的结构条件。`mx_vec_mult` 在矩阵有效且矩阵列数与向量长度一致时，逐行计算点积；输入类型不合适时抛出 `TypeError`，结构或维度不合适时抛出 `ValueError`。

## 更多向量的乐趣

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if isinstance(dimension, bool) or not isinstance(dimension, int):
            raise TypeError("维度必须是非负整数。")
        if dimension < 0:
            raise ValueError("维度不能为负数。")

        if elements is None:
            elements = (0,) * dimension
        elif not isinstance(elements, (list, tuple)):
            raise TypeError("向量元素必须由列表或元组提供。")
        elif len(elements) != dimension:
            raise ValueError("维度必须与元素个数相同。")

        if not all(_is_number(value) for value in elements):
            raise TypeError("向量元素必须是整数或浮点数。")

        self.dimension = dimension
        self.elements = tuple(elements)

    def _check_dimension(self, other):
        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须相同。")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_dimension(other)
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_dimension(other)
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_dimension(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_dimension(other)
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_dimension(other)
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("内积运算的参数必须是 Vector 对象。")
        self._check_dimension(other)
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, Vector):
            self._check_dimension(other)
            return Vector(
                self.dimension,
                tuple(a * b for a, b in zip(self.elements, other.elements)),
            )
        if _is_number(other):
            return Vector(
                self.dimension,
                tuple(value * other for value in self.elements),
            )
        return NotImplemented

    __rmul__ = __mul__


v1 = Vector(2, (2, 4))
v2 = Vector(2, [2, 5])
assert v1 < v2
assert Vector(2) == Vector(2, (0, 0))
assert v1.dot(v2) == 24
assert (2 * v1).elements == (4, 8)
assert (v1 * 2).elements == (4, 8)
assert (v1 * v2).elements == (4, 20)
print("v1 < v2：", v1 < v2)
print("默认零向量 Vector(2)：", Vector(2).elements)
print("v1 与 v2 的内积：", v1.dot(v2))
print("2 * v1：", (2 * v1).elements)
print("v1 * v2：", (v1 * v2).elements)
print("Vector 类测试通过。")
```

维度和元素都是实例属性，因为不同向量对象可以有不同维度和坐标；若设为类属性，不同对象会错误地共享这些数据。初始化时只给维度会得到全零向量，传入的列表或元组会被保存为元组。比较运算遵循坐标元组的字典序，并要求维度一致；`dot` 计算内积。乘法支持标量左乘、右乘和两个向量逐坐标相乘，结果均为新的 `Vector` 对象。测试用例覆盖了默认元素、排序、内积和两类乘法。
