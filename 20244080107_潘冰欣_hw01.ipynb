## 欧几里得算法

```python
def gcd(a, b):
    a = abs(a)
    b = abs(b)

    while b != 0:
        a, b = b, a % b

    return a


print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
print(gcd(-2018, 2019))
print(gcd(-1200, 300))
print(gcd(-5040, 60))
```

abs会把负数转成正数，因此这个实现对负数输入也能给出非负的最大公约数。

## 近似欧拉数e

```python
def euler_limit(n):
    return (1 + 1 / n) ** n

print(euler_limit(10))
print(euler_limit(100))
print(euler_limit(1000))
```

随着n增大，这个近似值会逐渐接近e

```python
import math

def euler_infinite_sum(n):
    total = 0.0
    for k in range(n):
        total += 1 / math.factorial(k)
    return total

print(euler_infinite_sum(4))
print(euler_infinite_sum(1))
print(euler_infinite_sum(0))
```

`euler_infinite_sum(n)` 将公式中前 `n` 项相加。这里从 `k = 0` 开始，因此 `euler_infinite_sum(4)` 计算的是 `1/0! + 1/1! + 1/2! + 1/3!`，结果约为 `2.6667`。`euler_infinite_sum(1)` 只计算第一项，结果为 `1.0`；`euler_infinite_sum(0)` 不计算任何项，结果为 `0.0`。

```python
import math

def euler_approx(epsilon):
    total = 0.0
    term = 1.0
    k = 0

    while math.exp(1) - total > epsilon:
        total += term
        k += 1
        term /= k

    return total


approximation = euler_approx(0.01)
print(f"近似值：{approximation}")
print(f"与 e 的差：{abs(math.exp(1) - approximation)}")
```

`euler_approx(0.01)` 使用级数中的各项逐步求和，直到近似值与 `math.exp(1)` 的差不超过 `0.01`。因此输出的近似值应接近 `2.718`，并且“与 e 的差”不大于 `0.01`。增大 `epsilon` 会较早停止，减小 `epsilon` 则会继续累加更多项。

```python
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"k={k}: {euler_infinite_sum(k)}")

print_euler_sum_table(5)
```

`print_euler_sum_table(5)` 依次计算并打印 `euler_infinite_sum(k)`，其中 `k` 从 1 到 5。每行显示当前的 `k` 和使用前 `k` 项得到的近似值。随着项数增加，近似值会逐渐接近 e（约为 `2.71828`）。

## 测试整数属性

```python
def is_power_of_2(n):
    while n % 2 == 0:
        n //= 2
    return n == 1


print(is_power_of_2(1))
print(is_power_of_2(8))
print(is_power_of_2(12))
```

`is_power_of_2(n)` 通过取余和整除不断除去因数 2。如果最后剩下 `1`，说明原数是 2 的幂，因此 `1` 和 `8` 返回 `True`；`12` 不能完全除尽因数 2，返回 `False`。

```python
def is_power(b, n):
    if b == 1:
        return n == 1

    while n % b == 0:
        n //= b

    return n == 1


print(is_power(2, 8))
print(is_power(3, 27))
print(is_power(2, 12))
print(is_power(1, 1))
print(is_power(1, 5))
```

`is_power(b, n)` 判断 `n` 是否为 `b` 的幂。函数不断将 `n` 除以 `b`；如果最后结果为 `1`，就返回 `True`。因此 `8` 是 `2` 的幂、`27` 是 `3` 的幂，而 `12` 不是 `2` 的幂。对于底数 `1`，只有 `n` 为 `1` 时返回 `True`。
