## 实验2：函数、条件、循环与递归

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在 0 到 100 之间。")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

for score in (0, 59, 60, 79, 80, 89, 90, 100):
    print(f"{score} 分：{score_level(score)}")
try:
    score_level(101)
except ValueError as error:
    print("越界分数检查：", error)

for_sum = 0
for number in range(1, 101):
    for_sum += number

while_sum = 0
number = 1
while number <= 100:
    while_sum += number
    number += 1
print(f"for 求和结果：{for_sum}:while 求和结果：{while_sum}；结果相同：{for_sum == while_sum}")

def check_factorial_input(n):
    if type(n) is not int:
        raise TypeError("阶乘输入必须是整数。")
    if n < 0:
        raise ValueError("阶乘输入必须是非负整数。")

def factorial_recursive(n):
    check_factorial_input(n)
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    check_factorial_input(n)
    result = 1
    for factor in range(2, n + 1):
        result *= factor
    return result

for n in (0, 1, 5, 10):
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)
    print(f"{n}!：递归={recursive_result}，迭代={iterative_result}，结果相同：{recursive_result == iterative_result}")

def summary(values):
    if not values:
        raise ValueError("values 不能为空。")
    return min(values), max(values), sum(values) / len(values)

minimum, maximum, average = summary([85, 58, 92, 76, 60, 89])
print(f"最小值：{minimum}，最大值：{maximum}，平均值：{average:.2f}")

def sum_all(*args):
    return sum(args)

print("sum_all() =", sum_all())
print("sum_all(7) =", sum_all(7))
print("sum_all(1, 2, 3, 4) =", sum_all(1, 2, 3, 4))
```

## 实验3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_by_name = dict(zip(names, scores))
print("姓名与成绩：")
for name, score in score_by_name.items():
    print(f"{name}: {score}")

print("序号、姓名与成绩：")
for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{index}. {name}: {score}")

passing_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print("及格学生：", passing_names)
print("成绩平方：", squared_scores)

student_count = len(scores)
average_score = sum(scores) / student_count
highest_score = max(scores)
lowest_score = min(scores)
pass_rate = len(passing_names) / student_count
print(f"人数：{student_count}，平均分：{average_score:.2f}，最高分：{highest_score}，最低分：{lowest_score}，及格率：{pass_rate:.2%}")

demo_scores = [85, 58, 92]
alias = demo_scores
copied = demo_scores.copy()
alias[0] = 100
print("alias 修改后原列表：", demo_scores)
print("alias:", alias)
print("独立副本：", copied)

student_record = ("Alice", 85)
print("学生记录元组：", student_record)
try:
    student_record[1] = 90
except TypeError as error:
    print(f"元组元素不能重新赋值：{error}")

short_zip = list(zip([1, 2, 3], ["甲", "乙"]))
print("长度不同的序列进行 zip:", short_zip, "（结果长度由较短序列决定）")
```

## 实验4：文件读写、异常处理与模块

```python
import importlib.util
import os
import pickle

work_dir = os.getcwd()
data_dir = os.path.join(work_dir, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
students_path = os.path.join(data_dir, "students.txt")
students_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""
with open(students_path, "w", encoding="utf-8") as file:
    file.write(students_text)
print("当前工作目录：", work_dir)
print("数据文件路径：", students_path)

def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError("字段数量必须为 2。")
    name = fields[0].strip()
    if not name:
        raise ValueError("姓名不能为空。")
    try:
        score = int(fields[1].strip())
    except ValueError as error:
        raise ValueError("成绩必须是整数。") from error
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间。")
    return {"name": name, "score": score}

valid_students = []
invalid_records = []
with open(students_path, "r", encoding="utf-8") as file:
    next(file, None)
    for line_number, line in enumerate(file, start=2):
        try:
            valid_students.append(parse_student(line))
        except ValueError as error:
            invalid_records.append({"line": line_number, "reason": str(error), "content": line.strip()})

print("有效记录：", valid_students)
print("无效记录：")
for record in invalid_records:
    print(f"第 {record['line']} 行：{record['content']}；原因：{record['reason']}")

valid_scores = [student["score"] for student in valid_students]
valid_count = len(valid_students)
average_valid_score = sum(valid_scores) / valid_count if valid_count else 0
valid_pass_rate = sum(score >= 60 for score in valid_scores) / valid_count if valid_count else 0
report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as file:
    file.write(f"有效记录数：{valid_count}\n")
    file.write(f"无效记录数：{len(invalid_records)}\n")
    file.write(f"有效记录平均分：{average_valid_score:.2f}\n")
    file.write(f"有效记录及格率：{valid_pass_rate:.2%}\n")
with open(report_path, "a", encoding="utf-8") as file:
    file.write("追加说明：本文件先以 w 模式创建，再以 a 模式追加本行。\n")
with open(report_path, "r", encoding="utf-8") as file:
    print("报告内容：\n" + file.read())

pickle_path = os.path.join(data_dir, "students.pkl")
with open(pickle_path, "wb") as file:
    pickle.dump(valid_students, file)
with open(pickle_path, "rb") as file:
    restored_students = pickle.load(file)
print("pickle 恢复数据与原数据相同：", restored_students == valid_students)

module_path = os.path.join(work_dir, "student_utils.py")
module_source = '''def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError("字段数量必须为 2。")
    name = fields[0].strip()
    if not name:
        raise ValueError("姓名不能为空。")
    try:
        score = int(fields[1].strip())
    except ValueError as error:
        raise ValueError("成绩必须是整数。") from error
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间。")
    return {"name": name, "score": score}
'''
with open(module_path, "w", encoding="utf-8") as file:
    file.write(module_source)
module_spec = importlib.util.spec_from_file_location("student_utils", module_path)
student_utils = importlib.util.module_from_spec(module_spec)
module_spec.loader.exec_module(student_utils)
print("模块有效输入：", student_utils.parse_student("Ivy,88"))
try:
    student_utils.parse_student("Jack,abc")
except ValueError as error:
    print("模块无效输入：", error)

missing_path = os.path.join(data_dir, "this_file_does_not_exist.txt")
try:
    with open(missing_path, "r", encoding="utf-8") as file:
        file.read()
except FileNotFoundError:
    print(f"文件不存在，已捕获异常：{missing_path}")
```

## 实验5：类、继承与对象复制

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"学生：{self.name}，成绩：{self.score}"

student_objects = [Student(record["name"], record["score"]) for record in valid_students]
for student in student_objects:
    print(f"{student}，及格：{student.is_passed()}")

class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"研究生：{self.name}，成绩：{self.score}，研究方向：{self.research_topic}"

graduate = GraduateStudent("Lin", 95, "数据分析")
print(graduate)
print("研究生及格：", graduate.is_passed())

original_direct = {"name": "Alice", "scores": [85, 90]}
direct_assignment = original_direct
direct_assignment["scores"].append(95)
print("直接赋值后，原对象：", original_direct)
print("直接赋值对象：", direct_assignment)

original_shallow = {"name": "Alice", "scores": [85, 90]}
shallow_copy = copy.copy(original_shallow)
shallow_copy["scores"].append(95)
print("浅拷贝后，原对象：", original_shallow)
print("浅拷贝对象：", shallow_copy)

original_deep = {"name": "Alice", "scores": [85, 90]}
deep_copy = copy.deepcopy(original_deep)
deep_copy["scores"].append(95)
print("深拷贝后，原对象：", original_deep)
print("深拷贝对象：", deep_copy)
```
